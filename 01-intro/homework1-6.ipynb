```python
med_old = car[car['model_year'] == 1996]['horsepower'].median()
med_new = car[car['model_year'] == 2014]['horsepower'].median()

if med_new > med_old:
    print(f"The median horsepower INCREASED from {med_old} to {med_new}.")
elif med_new < med_old:
    print(f"The median horsepower DECREASED from {med_old} to {med_new}.")
else:
    print(f"The median horsepower REMAINED THE SAME ({med_old}).")
```

```python

# 1. Get the median horsepower per year
yearly_medians = car.groupby('model_year')['horsepower'].median().to_frame()

# 2. Calculate the difference from the previous year
yearly_medians['diff'] = yearly_medians['horsepower'].diff()

# 3. Create text conditions based on the math
conditions = [yearly_medians['diff'] > 0, yearly_medians['diff'] < 0]
choices = ['Increased', 'Decreased']

# 4. Apply the text labels
yearly_medians['trend'] = np.select(conditions, choices, default='No Change / First Year')
yearly_medians
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Having a year-over-year median table
yearly_medians = car.groupby('model_year')['horsepower'].median().reset_index()

# yearly medians and add a linear trendline
plt.figure(figsize=(10, 5))
sns.regplot(
    data=yearly_medians, 
    x='model_year', 
    y='horsepower', 
    scatter_kws={'s': 50}, 
    line_kws={'color': 'red', 'linestyle': '--'} 
)

plt.title('Year-over-Year Median Horsepower Trend')
plt.xlabel('Model Year')
plt.ylabel('Median Horsepower')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
u = np.array([2, 4, 5, 6])
2 * u
```

```python
v = np.array([1, 0, 0, 2])
u + v
```

```python
u.shape[0]
```

```python
# A function that does vector multiplication
def vector_vector_multiplication(u, v):
    assert u.shape[0] == v.shape[0] # make sure the vectos have the same size, by changing the shape

    n = u.shape[0] # no big deal even if we write n = v.shape[0]

    result = 0.0
    
    for i in range(n):
        result = result + u[i] * v[i]

    return result
```

```python
vector_vector_multiplication(u,v).item()
```

```python
u.dot(v)
```

```python
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1],
])
```

```python
U.dot(v)
```

```python
def matrix_vector_multiplication(U, v):
    assert U.shape[1] == v.shape[0]
    
    num_rows = U.shape[0]
    
    result = np.zeros(num_rows)
    
    for i in range(num_rows):
        result[i] = vector_vector_multiplication(U[i], v)
    
    return result
```

```python
matrix_vector_multiplication(U, v)
```

```python
V = np.array([
    [1, 1, 2],
    [0, 0.5, 1], 
    [0, 2, 1],
    [2, 1, 0],
])
```

```python
def matrix_matrix_multiplication(U, V):
    assert U.shape[1] == V.shape[0]
    
    num_rows = U.shape[0]
    num_cols = V.shape[1]
    
    result = np.zeros((num_rows, num_cols))
    
    for i in range(num_cols):
        vi = V[:, i]
        Uvi = matrix_vector_multiplication(U, vi)
        result[:, i] = Uvi
    
    return result
```

```python
matrix_matrix_multiplication(U, V)
```

```python
U.dot(V)
```
